```python
import pandas as pd
import numpy as np
import re
```

```python
path = '/content/drive/MyDrive/crm_output_anonymized.csv'
```

```python
# =====================================================
# LOAD
# =====================================================
df = pd.read_csv(path, encoding="utf-8-sig", dtype=str)
df
out = df.copy()
```

```python
pd.set_option('display.max_columns', None)
out
```

```python
# =====================================================
# HELPERS
# =====================================================
def col(name):
    return df[name] if name in df.columns else pd.Series(np.nan, index=df.index)

def clean_phone(x):
    if pd.isna(x): return np.nan
    d = "".join(c for c in str(x) if c.isdigit())
    if d.startswith("380") and len(d) == 12: return "+" + d
    if d.startswith("0") and len(d) == 10: return "+38" + d
    return "+" + d if d else np.nan

clean_email = lambda x: np.nan if pd.isna(x) else str(x).strip().lower()

def clean_money(x):
    if pd.isna(x): return np.nan
    x = re.sub(r"[₴грн\s]", "", str(x)).replace(",", ".")
    x = re.sub(r"[^0-9.]", "", x)
    try: return float(x)
    except: return np.nan

def extract_phone(x):
    if pd.isna(x): return np.nan
    m = re.findall(r'[\+\d\-\(\)\s]{10,20}', str(x))
    return clean_phone(m[0]) if m else np.nan
```

```python
# =====================================================
# NOTES
# =====================================================
note_cols = [c for c in ["Note","Note 2","Note 3","Note 4","Note 5"] if c in df.columns]
out["All_Notes"] = df[note_cols].fillna("").agg(" ".join, axis=1)
```

```python
# =====================================================
# TELEGRAM
# =====================================================
def extract_telegram(x):
    if pd.isna(x): return np.nan
    for p in [r'@([A-Za-z0-9_]{5,32})', r't\.me/([A-Za-z0-9_]{5,32})']:
        m = re.search(p, str(x), re.I)
        if m: return "@" + m.group(1)
    return np.nan

out["Telegram"] = df['Telegram'].apply(extract_telegram)
```

```python
# =====================================================
# CONTACTS
# =====================================================
phone_notes = out["All_Notes"].apply(extract_phone)

out["Clean_Phone"] = (
    col("Mobile phone")
    .fillna(col("Work phone"))
    .fillna(col("Work DD phone"))
    .fillna(col("Home phone"))
    .fillna(col("Other phone"))
    .fillna(phone_notes)
    .apply(clean_phone)
)

out["Clean_Email"] = (
    col("Work email")
    .fillna(col("Personal email"))
    .fillna(col("Other email"))
    .apply(clean_email)
)
```

```python
# =====================================================
# COURSE
# =====================================================
out["Clean_Course"] = (
    col("Курс")
    .fillna(col("Lead title"))
    .fillna(col("COURSE"))
    .fillna(col("Название пакета"))
    .fillna(out["All_Notes"])
    .fillna("(no-course)")
)
```

```python
out["Clean_Sale"] = col("Lead sale ₴").apply(clean_money).fillna(0)
```

```python
# =====================================================
# UTM (compact merge)
# =====================================================
def merge(cols):
    return df.apply(
        lambda r: " | ".join(
            dict.fromkeys(
                str(r[c]).strip()
                for c in cols
                if c in df.columns and pd.notna(r[c]) and str(r[c]).strip() and str(r[c]).lower() != "nan"
            )
        ) or "(not set)",
        axis=1
    ).str.lower()


```

```python
out["UTM_Source"]   = merge(["utm_source","utm_source.1","TRAF_SRC","openstat_source"])
out["UTM_Medium"]   = merge(["utm_medium","TRAF_TYPE","utm_medium.1"])
out["UTM_Campaign"] = merge(["utm_campaign","utm_campaign.1","ADV_CAMP","openstat_campaign"])
out["UTM_Term"]     = merge(["utm_term","utm_term.1"])
out["UTM_Content"]  = merge(["utm_content","utm_content.1"])
out["Referrer"]     = merge(["Referrer","REFERER","referrer","utm_referrer"])
```

```python
utm_raw_cols = [
    "utm_source", "utm_source.1", "TRAF_SRC", "openstat_source",
    "utm_medium", "utm_medium.1", "TRAF_TYPE",
    "utm_campaign", "utm_campaign.1", "ADV_CAMP", "openstat_campaign",
    "utm_term", "utm_term.1",
    "utm_content", "utm_content.1",
    "REFERER", "referrer", "utm_referrer"
]

out = out.drop(columns=[c for c in utm_raw_cols if c in out.columns])
```

```python
# IDS + QUIZ
# =====================================================
out["Clean_Gclid"] = col("gclid").fillna(col("gclientid")).fillna(col("GOOGLE_ID"))

quiz_cols = [c for c in ["Вопрос 1","Вопрос 2","Вопрос 3","Вопрос 4","Вопрос 5"] if c in df.columns]

out["Quiz_Summary"] = df[quiz_cols].apply(
    lambda r: " | ".join(x.strip() for x in r if pd.notna(x) and str(x).strip()),
    axis=1
).replace("", np.nan)
```

```python
quiz_cols_to_drop = ["Вопрос 1","Вопрос 2","Вопрос 3","Вопрос 4","Вопрос 5"]
out = out.drop(columns=quiz_cols_to_drop)
```

```python
# =====================================================
# DATES
# =====================================================
out['Modified on'] = pd.to_datetime(out['Modified on'], dayfirst=True, errors='coerce')
out['Created on'] = pd.to_datetime(out['Created on'], dayfirst=True, errors='coerce')
closed = col("Closed at").replace("not closed", np.nan)
out["Closed_At"] = pd.to_datetime(
    closed,
    format="%d.%m.%Y %H:%M:%S",
    errors="coerce"
)
out['Year'] = out['Created on'].dt.year
out['Month number'] = out['Created on'].dt.month
out['Month name'] = out['Created on'].dt.month_name()
out['Week Number'] = out['Created on'].apply(lambda x: int(x.strftime('%W')) if pd.notna(x) else None)
```

```python
# =====================================================
# CLEANUP
# =====================================================
out = out.drop(columns=[c for c in df.columns if c not in out.columns and "utm" in c], errors="ignore")

def remove_garbage(df, thr=0.95):
    df = df.replace([r"^\s*$","nan","null","none"], np.nan, regex=True)
    return df.loc[:, ~(df.isna().mean() > thr)]

out = remove_garbage(out)

cols_to_drop = ["Note","Note 2","Note 3","Note 4","Note 5","Mobile phone","Work phone","Other phone","Work email","Personal email","Other email","COURSE","Название пакета","Lead sale ₴"]
out = out.drop(columns=cols_to_drop)

```

```python
# =====================================================
# EXPORT
# =====================================================
out.to_csv("powerbi_leads_clean.csv", index=False, encoding="utf-8-sig")

print("DONE:", out.shape)
```
